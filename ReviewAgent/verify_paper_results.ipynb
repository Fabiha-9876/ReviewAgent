# Verifying the paper's numerical claims

This notebook is a thin wrapper around `verify_paper_results.py`. It imports the script's
segments and runs them, so the notebook and the script can never disagree.

An earlier version of this notebook duplicated the verification logic inline. It drifted:
it kept recomputing the Stage-4 result the paper has since withdrawn (`+2.36` Likert,
`p < 0.0001`, from `experiments/exp2_human_eval.json`) and the condition-floored SpecCov
scores, both of which are retracted in `NOTICE_CORRECTION.md`. Wrapping the script instead
of copying it removes that failure mode.

Run from the repository root, after extracting `issuespec-data-bundle-v2.tar.gz`.


```python
import sys
from pathlib import Path

# Resolve the repo root from the notebook's own location, not a hardcoded path.
REPO = Path.cwd()
while not (REPO / 'verify_paper_results.py').exists() and REPO != REPO.parent:
    REPO = REPO.parent
assert (REPO / 'verify_paper_results.py').exists(), 'run this from inside the repository'
sys.path.insert(0, str(REPO))

import verify_paper_results as V
print(f'repo root: {REPO}')
print(f'{len(V.SEGMENTS)} segments available')
```
